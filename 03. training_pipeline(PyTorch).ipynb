```python
import pandas as pd

# Load dataset from URL into a pandas DataFrame
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')

# Display first 5 rows of the DataFrame to understand its structure and data
df.head()
```

### train test split

```python
from sklearn.model_selection import train_test_split

# Split the dataset into features (X) and target (y)
# Features: all columns except the first (diagnosis column is assumed first)
# Target: first column (diagnosis or label)

# Split data into training and testing sets with 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2, random_state=42)

# Print shapes to verify split
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

### scaling

```python
from sklearn.preprocessing import StandardScaler

# Initialize the StandardScaler to normalize features
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train = scaler.fit_transform(X_train)

# Transform the test data using the same scaler (do NOT fit again)
X_test = scaler.transform(X_test)

# Print shape info to confirm transformation
print(f"Scaled X_train shape: {X_train.shape}")
print(f"Scaled X_test shape: {X_test.shape}")
```

### Label Encoding

```python
from sklearn.preprocessing import LabelEncoder

# Initialize LabelEncoder to convert categorical target labels to integers
encoder = LabelEncoder()

# Fit encoder on training labels and transform them
y_train = encoder.fit_transform(y_train)

# Transform test labels using the same encoder
y_test = encoder.transform(y_test)

# Print unique classes and sample encoded labels to verify
print(f"Classes: {encoder.classes_}")
print(f"Encoded y_train sample: {y_train[:5]}")
print(f"Encoded y_test sample: {y_test[:5]}")
```

### Numpy arrays to PyTorch tensors

```python
import torch

# Convert numpy arrays to PyTorch tensors
X_train_tensor = torch.from_numpy(X_train).double()  # Convert features to float tensors
X_test_tensor = torch.from_numpy(X_test).double()
y_train_tensor = torch.from_numpy(y_train).long()  # Convert labels to long tensors (for classification)
y_test_tensor = torch.from_numpy(y_test).long()

# Print shapes and types to confirm
print(f"X_train_tensor shape: {X_train_tensor.shape}, dtype: {X_train_tensor.dtype}")
print(f"X_test_tensor shape: {X_test_tensor.shape}, dtype: {X_test_tensor.dtype}")
print(f"y_train_tensor shape: {y_train_tensor.shape}, dtype: {y_train_tensor.dtype}")
print(f"y_test_tensor shape: {y_test_tensor.shape}, dtype: {y_test_tensor.dtype}")
```

### Defining the model

```python
import torch

class MySimpleNN():
    def __init__(self, X):
        # Initialize weights with random values
        # Shape: (number_of_features, 1), dtype float64 for precision, track gradients
        self.weights = torch.rand(X.shape[1], 1, dtype=torch.float64, requires_grad=True)
        # Initialize bias as zero, scalar, dtype float64, track gradients
        self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)


    def forward(self, X):
        # Linear combination: z = X * weights + bias
        z = torch.matmul(X, self.weights) + self.bias
        # Apply sigmoid activation to get probability predictions
        y_pred = torch.sigmoid(z)
        return y_pred

    def loss_function(self, y_pred, y):
        # Small constant to prevent log(0) errors in loss calculation
        epsilon = 1e-7
        # Clamp predictions to range [epsilon, 1 - epsilon]
        y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
        # Compute binary cross-entropy loss (mean over batch)
        loss = -(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred)).mean()
        return loss
```

### Important Parameters

```python
# Set learning rate for gradient descent updates
learning_rate = 0.1

# Set number of epochs (full passes over training data)
epochs = 25
```

### Training Pipeline

```python
# Create model instance with training data shape
model = MySimpleNN(X_train_tensor)

# Training loop for specified number of epochs
for epoch in range(epochs):
    # Forward pass: compute predicted outputs by passing inputs to the model
    y_pred = model.forward(X_train_tensor)

    # Compute the loss between predicted outputs and true labels
    loss = model.loss_function(y_pred, y_train_tensor)

    # Backward pass: compute gradients of the loss w.r.t. model parameters
    loss.backward()

    # Update model parameters using gradient descent inside torch.no_grad() context
    # to prevent tracking these operations in the computation graph
    with torch.no_grad():
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

    # Zero the gradients after updating weights to avoid accumulation
    model.weights.grad.zero_()
    model.bias.grad.zero_()

    # Print current epoch number and loss value for monitoring
    print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

### Evaluation

```python
# Evaluate the model on test data without tracking gradients
with torch.no_grad():
    # Forward pass to get predictions
    y_pred = model.forward(X_test_tensor)

    # Convert probabilities to binary predictions using threshold 0.5 (commonly used)
    y_pred = (y_pred > 0.5).float()

    # Calculate accuracy by comparing predictions with true labels
    accuracy = (y_pred.squeeze() == y_test_tensor).float().mean()

    print(f'Accuracy: {accuracy.item()}')
```
