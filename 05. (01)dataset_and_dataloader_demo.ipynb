```python
# Import function to generate synthetic classification datasets
from sklearn.datasets import make_classification

# Import PyTorch for tensor computations and model building
import torch
```

```python
# Step 1: Create a synthetic classification dataset using sklearn
X, y = make_classification(
    n_samples=10,       # Total number of samples in the dataset
    n_features=2,       # Number of features per sample
    n_informative=2,    # Number of informative features contributing to the class label
    n_redundant=0,      # Number of redundant features (linear combinations of informative features)
    n_classes=2,        # Number of distinct classes (binary classification)
    random_state=42     # Seed for reproducibility of dataset generation
)
```

```python
# Convert the feature matrix to a PyTorch tensor with float32 dtype (standard for inputs)
X = torch.tensor(X, dtype=torch.float32)

# Convert the labels to a PyTorch tensor with long dtype (required for classification targets)
y = torch.tensor(y, dtype=torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
from torch.utils.data import Dataset, DataLoader

class CustomDataset(Dataset):
    def __init__(self, features, labels):
        # Store features and labels tensors
        self.features = features
        self.labels = labels

    def __len__(self):
        # Return total number of samples in the dataset
        return self.features.shape[0]

    def __getitem__(self, index):
        # Retrieve feature-label pair at the given index
        return self.features[index], self.labels[index]
```

```python
dataset = CustomDataset(X, y)
```

```python
dataloader = DataLoader(dataset, batch_size=2, shuffle=False)
```
