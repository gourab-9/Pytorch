```python
# Numpy for numerical operations
import numpy as np

# Pandas for data loading, cleaning, and manipulation
import pandas as pd

# PyTorch for building and training neural networks
import torch

# Split data into training and testing sets
from sklearn.model_selection import train_test_split

# Standardize features by removing the mean and scaling to unit variance
from sklearn.preprocessing import StandardScaler

# Encode categorical labels into numeric format (e.g., 'M'/'B' to 1/0)
from sklearn.preprocessing import LabelEncoder
```

```python
# Load the dataset from the GitHub URL into a Pandas DataFrame
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')

# Display the first five rows of the dataset to inspect structure and values
df.head()
```

```python
# Check the number of rows and columns in the DataFrame
df.shape
```

```python
# Drop the 'id' column (not useful for training) and the 'Unnamed: 32' column (all NaNs)
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
df.head()
```

### train test split

```python
# Split the dataset into features and labels
# Features: all columns from index 1 onward (exclude the 'diagnosis' column at index 0)
# Labels: the 'diagnosis' column (index 0)
# test_size=0.2 means 20% data for testing, 80% for training
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

### scaling

```python
# Initialize the StandardScaler for feature normalization
scaler = StandardScaler()

# Fit the scaler on the training features and transform them
# This removes the mean and scales to unit variance based on training data statistics
X_train = scaler.fit_transform(X_train)

# Transform the test features using the same scaler fitted on training data
# Ensures test data is scaled consistently without data leakage
X_test = scaler.transform(X_test)
```

### Label Encoding

```python
# Initialize LabelEncoder to convert categorical labels into numeric format
encoder = LabelEncoder()

# Fit the encoder on training labels and transform them into numeric values
# Example: 'M' (malignant) -> 1, 'B' (benign) -> 0
y_train = encoder.fit_transform(y_train)

# Transform the test labels using the same encoding learned from training labels
y_test = encoder.transform(y_test)
```

```python
y_train
```

### Numpy arrays to PyTorch tensors

```python
# Convert NumPy arrays to PyTorch tensors with dtype float32 for compatibility

# Features tensors (inputs) - converting scaled training and testing data
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))

# Labels tensors (targets) - converting encoded training and testing labels
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
X_train_tensor.shape
```

```python
y_train_tensor.shape
```

### Defining the model

```python
import torch.nn as nn

class MySimpleNN(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        # Linear layer: maps input features to a single output (logit)
        self.linear = nn.Linear(num_features, 1)
        # Sigmoid activation to squash output to [0, 1], interpretable as probability
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        # Compute linear combination
        out = self.linear(features)
        # Apply sigmoid activation
        out = self.sigmoid(out)
        return out
```

### Important Parameters

```python
# Learning rate controls the size of steps during gradient descent
learning_rate = 0.1

# Number of times the entire training dataset will be passed through the model
epochs = 25
```

```python
# Define the Binary Cross-Entropy loss function for binary classification
loss_function = nn.BCELoss()
```

### Training Pipeline

```python
# Create the model instance with the number of input features
model = MySimpleNN(X_train_tensor.shape[1])

# Define the optimizer: Stochastic Gradient Descent with specified learning rate
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# Training loop running for the specified number of epochs
for epoch in range(epochs):

    # Forward pass: compute predicted outputs by passing inputs to the model
    y_pred = model(X_train_tensor)

    # Calculate the loss between predicted outputs and true labels
    # y_train_tensor.view(-1,1) reshapes labels to match output shape (batch_size, 1)
    loss = loss_function(y_pred, y_train_tensor.view(-1, 1))

    # Clear previous gradients from the optimizer
    optimizer.zero_grad()

    # Backward pass: compute gradients of the loss with respect to model parameters
    loss.backward()

    # Update model parameters based on computed gradients
    optimizer.step()

    # Print the loss value for monitoring training progress
    print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

### Evaluation

```python
# Disable gradient computation for evaluation (speeds up computation and saves memory)
with torch.no_grad():
    # Forward pass on test data to get predictions
    y_pred = model.forward(X_test_tensor)

    # Convert probabilities to binary predictions (threshold = 0.5)
    y_pred = (y_pred > 0.5).float()

    # Calculate accuracy by comparing predictions with true labels
    accuracy = (y_pred == y_test_tensor).float().mean()

    # Print the accuracy score
    print(f'Accuracy: {accuracy.item()}')
```
