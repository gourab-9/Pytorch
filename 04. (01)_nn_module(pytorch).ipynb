```python
# Import necessary PyTorch modules
import torch
import torch.nn as nn

# Define a custom neural network model by subclassing nn.Module
class Model(nn.Module):

    def __init__(self, num_features):
        super().__init__()  # Initialize the base class

        # Define the network architecture using nn.Sequential for cleaner code
        self.network = nn.Sequential(
            nn.Linear(num_features, 3),  # First layer: fully connected (input -> 3 neurons)
            nn.ReLU(),                   # Activation function: ReLU
            nn.Linear(3, 1),             # Second layer: 3 neurons -> 1 output
            nn.Sigmoid()                 # Sigmoid to output probability (for binary classification)
        )

    def forward(self, features):
        # Define the forward pass
        out = self.network(features)
        return out
```

```python
# Create a random dataset with 10 samples and 5 features each
features = torch.rand(10, 5)

# Instantiate the model by passing the number of features (5 in this case)
model = Model(features.shape[1])

# Perform a forward pass through the model
# You can call model(features) directly instead of model.forward(features)
output = model(features)

# Print the output probabilities from the model
print(output)
```

```python
# Access the second Linear layer (index 2 in Sequential)
print(model.network[2].weight)  # This is Linear(3, 1)
```

```python
import torch
import torch.nn as nn

class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()

        # Define layers with explicit names for easy access
        self.linear1 = nn.Linear(num_features, 3)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        x = self.linear1(features)
        x = self.relu(x)
        x = self.linear2(x)
        out = self.sigmoid(x)
        return out
```

```python
# Create dummy input
features = torch.rand(10, 5)

# Create model
model = Model(features.shape[1])

# Forward pass
output = model(features)

# Access weights of second linear layer
print(model.linear2.weight)
```
